```python
import sys, os, warnings
import numpy as np
import pandas as pd
import xarray as xr
from datetime import datetime

import src.steps as steps
from src.utils.post_process import merge_bursts
from src.burst_detection.numba.exhaustive_search import compute_all_bursts, compute_all_bursts_numba_abs_incomplete_years
import utils_notebook as utils

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import matplotlib.cm as cm
import matplotlib.pyplot as plt
from matplotlib.colors import BoundaryNorm

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler


import pymannkendall as mk


sys.path.insert(0,'.')
import libimp
```

```python
start_year = 1974
end_year = 2023
n_year = end_year - start_year + 1
len_year = 92 * 24

ds = xr.open_dataset("../data/wgbt_era5_Paris_1974-2023jja.nc")
ts_wbgt = ds.wbgt.values.ravel().astype(np.float64)
ts_date = pd.DatetimeIndex(ds.time.values)
ts_tas = xr.open_dataset("../data/ta_era5_Paris_1974-2023jja.nc").t2m.values.ravel().astype(np.float64) - 273.15

ts_mort = np.load("mortality.npy").astype(np.float64)[:-92]

mask = (ts_date.year >= start_year) & (ts_date.year <= end_year)
ts_wbgt = ts_wbgt[mask]
ts_tas = ts_tas[mask]
ts_date = ts_date[mask]

tas_yearly = np.copy(ts_tas.reshape((n_year, len_year)))
wbgt_yearly = np.copy(ts_wbgt.reshape((n_year, len_year)))
mort_yearly = np.copy(ts_mort.reshape((n_year, len_year//24)))
```

```python
threshold = 3
minimum_length = 3
maximum_length = 10*24
```

```python
tas_bursts = compute_all_bursts_numba_abs_incomplete_years(ts_tas, n_year, threshold, len_year, minimum_length, maximum_length)
tas_bursts_merged = merge_bursts(tas_bursts)

wbgt_bursts = compute_all_bursts_numba_abs_incomplete_years(ts_wbgt, n_year, threshold, len_year, minimum_length, maximum_length)
wbgt_bursts_merged = merge_bursts(wbgt_bursts)
```

```python
tas_df, tas_yearly_sum = utils.compute_statistics(ts_tas, ts_date, tas_bursts_merged, len_year)
wbgt_df, wbgt_yearly_sum = utils.compute_statistics(ts_wbgt, ts_date, wbgt_bursts_merged, len_year)
```

```python
colors = {
'Hot': '#FFDDC1',  # Light red for Hot bursts
'Cold': '#C1D3FF'  # Light blue for Cold bursts
}
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=[
        "Temperature Bursts",
        "WBGT Bursts"
    ],
    specs=[[{"type": "table"},
            {"type": "table"}]],
)
cell_colors = [[colors.get(value, "white") for value in tas_df["Type"]] for _ in tas_df.columns]
fig.add_trace(
    go.Table(
        header=dict(
            values=["<b>Start Date</b>", "<b>End Date</b>", "<b>Length</b>", "<b>Type</b>"],
            fill_color="lightgrey",
            align="center",
            font=dict(color="black", size=14)
        ),
        cells=dict(
            values=[tas_df[col] for col in tas_df.columns],
            fill_color=cell_colors,
            align="center",
            font=dict(color="black", size=12)
        )
    ),
    row=1, col=1
)
cell_colors = [[colors.get(value, "white") for value in wbgt_df["Type"]] for _ in wbgt_df.columns]
fig.add_trace(
    go.Table(
        header=dict(
            values=["<b>Start Date</b>", "<b>End Date</b>", "<b>Length</b>", "<b>Type</b>"],
            fill_color="lightgrey",
            align="center",
            font=dict(color="black", size=14)
        ),
        cells=dict(
            values=[wbgt_df[col] for col in wbgt_df.columns],
            fill_color=cell_colors,
            align="center",
            font=dict(color="black", size=12)
        )
    ),
    row=1, col=2
)
fig.update_layout(
    width=1100,
    height=500,
    title_x=0.5,
    title_text = "Tables of burst information for both temperature and WBGT",

)
fig.write_html("double_table.html")
fig.show()
```

```python
overlapping_anomalies_df, tas_excl_df, wbgt_excl_df = utils.compute_bursts_stats(tas_bursts_merged, wbgt_bursts_merged, ts_date, "Temperature", "WBGT")
```

```python
fig = make_subplots(
    rows=2, cols=2,
    horizontal_spacing = 0.01,
    vertical_spacing = 0.005,
    subplot_titles=[
        "Overlapping Bursts",
        "Exclusive Temperature Bursts",
        "Exclusive WBGT Bursts", 
    ],
    specs=[
        [{"type": "table", "colspan": 2}, None],
        [{"type": "table"}, {"type": "table"}]
    ]
)
fig.add_trace(
    go.Table(
        header=dict(
            values=["<b>Start Date</b>", "<b>End Date</b>", "<b>Length</b>"],
            fill_color="lightgrey",
            align="center",
            font=dict(color="black", size=14)
        ),
        cells=dict(
            values=[tas_excl_df[col] for col in tas_excl_df.columns],
            fill_color = ['#FFDDB1', '#FFDDB1', '#FFDDB1'],
            align="center",
            font=dict(color="black", size=12)
        )
    ),
    row=2, col=1
)
fig.add_trace(
    go.Table(
        header=dict(
            values=["<b>Start Date</b>", "<b>End Date</b>", "<b>Length</b>",],
            fill_color="lightgrey",
            align="center",
            font=dict(color="black", size=14)
        ),
        cells=dict(
            values=[wbgt_excl_df[col] for col in wbgt_excl_df.columns],
            fill_color = [ '#FFDDE1', '#FFDDE1', '#FFDDE1'],
            align="center",
            font=dict(color="black", size=12)
        )
    ),
    row=2, col=2
)
if not overlapping_anomalies_df.empty:
    fig.add_trace(
        go.Table(
            header=dict(
                values=[
                    "<b>Overlap Start</b>", "<b>Overlap End</b>", "<b>Overlap Length</b>",
                    "<b>Temperature Start</b>", "<b>Temperature End</b>", "<b>Temperature Length</b>",
                    "<b>WBGT Start</b>", "<b>WBGT End</b>", "<b>WBGT Length</b>"                    
                ],
                fill_color="lightgrey",
                align="center",
                font=dict(color="black", size=14)
            ),
            cells=dict(
                values=[overlapping_anomalies_df[col] for col in overlapping_anomalies_df.columns],
                fill_color = ['#FFDDD1', '#FFDDD1', '#FFDDD1', '#FFDDB1', '#FFDDB1', '#FFDDB1', '#FFDDE1', '#FFDDE1', '#FFDDE1'],
                align="center",
                font=dict(color="black", size=12)
            )
        ),
        row=1, col=1
    )
fig.update_layout(
    width=1100,
    height=700,
    title_x=0.5,
    title_text = "Tables of overlapping and exclusive burst information for both temperature and WBGT",

)
fig.write_html("triple_table.html")

fig.show()
```

```python
fig = make_subplots(rows=2, cols=1,vertical_spacing = 0.05, shared_xaxes=True,
                    subplot_titles=("Temperature", "WBGT"))

hover_text = [f"{ts_date[i].strftime("%Hh, %d %b")}" for i in range(0, len_year)]  
fig.add_trace(go.Scatter(
    y=np.mean(tas_yearly, axis=0),
    mode='lines',
    line=dict(color="black", dash="dash"),
    name="Daily average",
    legendgroup="tas",  
    legendgrouptitle_text="Temperature",
    hovertext=hover_text),
    row=1, col=1
)
fig.add_trace(go.Scatter(
    y=np.mean(tas_yearly, axis=0) + 3*np.std(tas_yearly, axis=0),
    mode='lines',
    line=dict(color="black", dash="dot"),
    name="Daily threshold",
    legendgroup="tas", 
    hovertext=hover_text),
    row=1, col=1
)
for year in range(n_year):
    hover_text = [f"{ts_date[i].strftime("%Hh, %d %b %Y")}" for i in range(year*len_year, (year+1)*len_year)]  
    fig.add_trace(go.Scatter(
        y=tas_yearly[year, :],
        mode='lines',
        line=dict(color="grey", dash="dot", width=1),
        opacity=0.4,
        name=f"{start_year + year}",
        showlegend=False,
        hovertext=hover_text),
        row=1, col=1
    )
for (start, end, z) in tas_bursts_merged:
    text = "Hot" if z>0 else "Cold"
    length = end - start + 1
    i_start = start % len_year
    year = start // len_year
    xs = np.arange(i_start, i_start + length)
    hover_text = [f"{ts_date[i].strftime("%Hh, %d %b %Y")}" for i in range(start, end+1)]  
    fig.add_trace(go.Scatter(
        x=xs,
        y=ts_tas[start:start + length],
        mode='lines',
        line=dict(width=2.5),
        name=f"{start_year + year} {text} Burst",
        hovertext= hover_text,
        legendgroup="tas",  ),
        row=1, col=1
    )
hover_text = [f"{ts_date[i].strftime("%Hh, %d %b")}" for i in range(0, len_year)]  
fig.add_trace(go.Scatter(
    y=np.mean(wbgt_yearly, axis=0),
    mode='lines',
    line=dict(color="black", dash="dash"),
    name="Daily average",
    legendgroup="wbgt", 
    legendgrouptitle_text="WBGT",
    hovertext = hover_text),
    row=2, col=1
)
fig.add_trace(go.Scatter(
    y=np.mean(wbgt_yearly, axis=0)+3*np.std(wbgt_yearly, axis=0),
    mode='lines',
    line=dict(color="black", dash="dot"),
    name="Daily threshold",
    legendgroup="wbgt",
    hovertext = hover_text),
    row=2, col=1
)
for year in range(n_year):
    hover_text = [f"{ts_date[i].strftime("%Hh, %d %b %Y")}" for i in range(year*len_year, (year+1)*len_year)]  
    fig.add_trace(go.Scatter(
        y=wbgt_yearly[year, :],
        mode='lines',
        line=dict(color="grey", dash="dot", width=1),
        opacity=0.4,
        name=f"{start_year + year}",
        showlegend=False,
        hovertext = hover_text),
        row=2, col=1
    )

for (start, end, z) in wbgt_bursts_merged:
    text = "Hot" if z>0 else "Cold"
    length = end - start + 1
    i_start = start % len_year
    year = start // len_year
    xs = np.arange(i_start, i_start + length)
    hover_text = [f"{ts_date[i].strftime("%Hh, %d %b %Y")}" for i in range(start, end+1)]  
    fig.add_trace(go.Scatter(
        x=xs,
        y=ts_wbgt[start:start + length],
        mode='lines',
        line=dict(width=2.5),
        name=f"{start_year + year} {text} Burst",
        legendgroup="wbgt",
        hovertext= hover_text),
        row=2, col=1
    )

first_day_of_months = np.where(ts_date[:len_year].day == 1)[0][::24]
fifteen_day_of_months = np.where(ts_date[:len_year].day == 15)[0][::24]

days = [None]*(len(first_day_of_months)+len(fifteen_day_of_months))
days[::2] = first_day_of_months
days[1::2] = fifteen_day_of_months

fig.update_xaxes(
    tickvals=days,
    ticktext=ts_date[days].strftime("%d %B"),
    tickangle=0
)

fig.update_layout(height=700, width=1100,    title_text = "Time series with bursts (in color) for both temperature and WBGT",
)
# Show the plot
fig.write_html("double_series.html")
fig.show()
```
